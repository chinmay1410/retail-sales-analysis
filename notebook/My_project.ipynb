```python
import numpy as np
import pandas as pd


df = pd.read_csv('/content/Sample - Superstore.csv.zip', encoding = 'latin-1')

print(df.shape)                             #row and columns
df.head()                                   #first 5 rows
df.info()                                   #data types and non-null counts
df.describe()                               #min, max, mean of numeric columns
print(df.isnull().sum())                    #missing values per column
print(df.duplicated().sum())                #duplicate rows
```

```python
# Check your columns names!
df.columns
```

```python
# Standardize columns names: lowercase and hyphens with underscores!
df.columns = (c.lower().replace(" ", "_").replace("-", "_") for c in df.columns)
df.columns
```

```python
# Convert dates from to texts to real dates!
df['order_date'] = pd.to_datetime(df['order_date'])
df['ship_date'] = pd.to_datetime(df['ship_date'])

print(df['order_date'].head()) #Check that days and months were read correctly!

# Remove duplicate rows!
before = len(df)
df = df.drop_duplicates()
print('\nDuplicates Removed: ', before - len(df))
```

```python
# Trim spaces in text columns!
for col in ['segment','region','category', 'sub_category','state','city']:
  df[col] = df[col].str.strip()

# Postal code is an identifier, not a number!
df['postal_code'] = df['postal_code'].astype(str)

# Create useful new columns!
df['order_year'] = df['order_date'].dt.year
df['order_month'] = df['order_date'].dt.month
df['ship_days'] = (df['ship_date'] - df['order_date']).dt.days
df['profit_margin_pct'] = (df['profit'] / df['sales'] * 100).round(2)
```

```python
# Sanity checks!
print((df['sales'] <= 0).sum(), 'rows with non-positive sales.')
print((df['ship_days'] < 0).sum(), 'rows shipped before ordered.')
print(df[['sales','quantity','discount','profit']].describe())

# Save the cleaned file!
df.to_csv('/content/Sample - Superstore_cleaned.csv.zip', index = False)

#download the cleaned file!
from google.colab import files
files.download('/content/Sample - Superstore_cleaned.csv.zip')
```
